## Функции

*Аналогия 1 (кухня)*: функция - это рецепт. Вы даёте ингредиенты, выполняете шаги и получаете блюдо.

*Аналогия 2 (завод)*: функция - это станок. Кладёте заготовку, станок её обрабатывает, на выходе - готовая деталь.

* `def` - «объявляю рецепт»;

* `()` - «какие ингредиенты принимаю»;

* `return` - «что выдаю обратно».

Когда мы делаем LLM или даже просто готовим тексты, одни и те же куски повторяются постоянно:

* прочитать файл;

* почистить строку;

Если писать это каждый раз заново - получится каша. Функции дают три большие выгоды:

* Не повторяемся - один рецепт, готовим по нему много раз.

* Код читаемый - имя функции работает как заголовок в книге.

* Легче тестировать - проверили рецепт на паре примеров и уверены в нём.

Пусть задача - почистить текст

```python
def clean_text(s):
    s = s.strip()          # убрать пробелы по краям
    s = s.lower()          # сделать маленькими
    s = s.replace("!", "") # убрать !
    return s

print(clean_text("  Привет!!!  "))  # "привет"
```

## Работа с файлами и текстом

LLM - это «двигатель», который учится на огромном количестве текста. На практике ваша программа делает три шага, как на кухне:

* Достать продукты - прочитать файлы;

* Помыть и почистить - нормализовать текст;

* Нарезать - разбить на куски, строки, токены.

Файл - это текст на диске, как книга на полке. Чтобы прочитать «книгу», её нужно открыть, прочитать и закрыть.

В Python самый правильный способ - через `with`: это как «взял книгу, прочитал, положил обратно» автоматически.

```python
with open("data.txt", "r", encoding="utf-8") as f:
    text = f.read()

print(text[:100])  # первые 100 символов
```

* `"r"` - режим чтения (read);

* `encoding="utf-8"` - чтобы русский текст не ломался;

* `f.read()` - читает всё сразу (подходит для небольших файлов).

Если файл большой, читать его целиком тяжело. Тогда читаем по строкам:

```python
with open("data.txt", "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if line:
            print(line)
```

Перезаписать файл:

```python
with open("out.txt", "w", encoding="utf-8") as f:
    f.write("Привет!\n")
    f.write("Это новая строка.\n")
```

`"w"` - перезапишет файл заново.

Дописать в конец (лог, результаты):

```python
with open("out.txt", "a", encoding="utf-8") as f:
    f.write("Ещё одна строка\n")
```

`"a"` - append (добавить в конец).

### Мини-пайплайн подготовки текста

Упрощённый «конвейер»:

**вход:** текст из файла. 

**выход:** список чистых строк без пустых.

```python
def clean_lines(path):
    out = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip().lower()
            line = line.replace("!", "").replace("?", "")
            if line:              # пропускаем пустые
                out.append(line)
    return out
```

### Самое главное - autograd!
Именно это отличает Pytorch от NumPy

**PyTorch умеет автоматически считать производные!**

Это значит, что он сам понимает, как менять веса, чтобы ошибка уменьшалась. И всё это делает за нас - великолепно же.



```python
import torch

w = torch.tensor(2.0, requires_grad=True)
x = torch.tensor(3.0)

y = w * x
y.backward()

print(w.grad)
```

Почему 3? Потому что y = w * x, и при x = 3 производная по w равна 3.

Без autograd производные пришлось бы считать вручную! Для LLM это тысячи операций - руками невозможно. Autograd делает это автоматически.
